## Carga de datos

Ejecuta la siguiente celda. No requiere modificaciones.

```python
import pandas as pd

# Carga del dataset de taxis de Nueva York desde un repositorio público de GitHub.
# El archivo se lee con pandas; no se requiere ninguna otra biblioteca.
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/taxis.csv'
taxis = pd.read_csv(url)

print('Dimensiones:', taxis.shape)
taxis.head()
```

## Actividad 1. Conversión y validación (20 puntos)

### 1.1 Conversión (5 pts)

1. Muestra el tipo de dato de `pickup` y `dropoff`.
2. Convierte ambas columnas a tipo fecha con `pd.to_datetime()`, indicando el **formato explícito** que corresponde a los valores.
3. Verifica que la conversión no generó `NaT`: compara el número de faltantes antes y después.

```python
print(taxis[['pickup', 'dropoff']].dtypes) # mostramos los tipos de datos

na_antes = taxis[['pickup', 'dropoff']].isna().sum() #primer conteo de faltantes antes de la conversion

taxis['pickup'] = pd.to_datetime(taxis['pickup'], format='%Y-%m-%d %H:%M:%S') #convertimos ambas columnas a formato de fecha
taxis['dropoff'] = pd.to_datetime(taxis['dropoff'], format='%Y-%m-%d %H:%M:%S')

na_despues = taxis[['pickup', 'dropoff']].isna().sum() #conteo de faltantes despues de la conversion
print("\nFaltantes antes:\n", na_antes)
print("\nFaltantes después:\n", na_despues)
```

## Actividad 2. Duraciones y anomalías (25 puntos)

### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.
2. Crea `duracion_min` con la duración en **minutos** (usa `.dt.total_seconds()`).
3. Muestra el resumen estadístico de `duracion_min` con `describe()`.

```python
taxis['duracion'] = taxis['dropoff'] - taxis['pickup'] #nueva columna
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60 #ajustamos a minutos

print(taxis['duracion_min'].describe())
```

### 2.2 Duraciones anómalas (6 pts)

Cuenta los viajes con:

- Duración **negativa** (termina antes de iniciar).
- Duración **cero**.
- Duración **menor a 1 minuto** (incluye los de duración cero).

Muestra los viajes de duración cero con las columnas `pickup`, `dropoff`, `distance`, `fare` y `total`.

```python
print("Duracion negativa: ",(taxis['duracion_min'] < 0).sum()) #aquellos menores a 0
print("Duración cero:",(taxis['duracion_min'] == 0).sum()) #exatamente igual que 0
print(f"Duración menor a 1 min (incluye cero): {(taxis['duracion_min'] < 1).sum()}") # menores a 1 minuto
print(f"Duración mayor a 3 horas: {(taxis['duracion_min'] > 180).sum()}") #mayor a 3 horas
print("\nRegistros de duracion cero:")
print(taxis[taxis['duracion_min'] == 0][['pickup', 'dropoff', 'distance', 'fare', 'total']]) #viajes con solamente la duracion 0
#print(taxis[taxis['duracion_min'] < 1][['pickup', 'dropoff', 'distance', 'fare', 'total']])
```

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
# para crear la magnitud velocidad dividimos distancia entre tiempo
taxis['velocidad_mph'] = (taxis['distance'] / (taxis['duracion_min'] / 60)).where(taxis['duracion_min'] > 0)

# contamos aquellos viajes los cuales la distancia es mayor a 60 mph
print(f"Viajes con velocidad > 60 mph: {(taxis['velocidad_mph'] > 60).sum()}")

print("\nTop 5 viajes con mas velocidad:")
top_vel = taxis.sort_values(by='velocidad_mph', ascending=False).head(5) # ordenamos por velocidad y mostramos las 5
print(top_vel[['pickup', 'dropoff', 'duracion_min', 'distance', 'velocidad_mph', 'fare']])
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
taxis['hora'] = taxis['pickup'].dt.hour #hora
taxis['dia_num'] = taxis['pickup'].dt.dayofweek #semana
taxis['dia_nombre'] = taxis['pickup'].dt.day_name() # nombre del dia
taxis['fin_semana'] = taxis['dia_num'] >= 5 # es sabado o domingo
taxis['fecha'] = taxis['pickup'].dt.date #fecha
taxis['cruza_medianoche'] = taxis['pickup'].dt.date != taxis['dropoff'].dt.date #el viaje termino al siguiente dia

print(f"Viajes que cruzan la medianoche: {taxis['cruza_medianoche'].sum()}")
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
#se agrupan los viajes por hora, contamos cada uno y obtenemos su mediana
horario = taxis.groupby('hora').agg(viajes=('hora', 'size'),mediana_duracion=('duracion_min', 'median'))
print(horario)
print(f"\nHora con mas viajes: {horario['viajes'].idxmax()}:00")
print(f"Hora con menos viajes: {horario['viajes'].idxmin()}:00")
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
#agrupamos por dia de la semana calculando sus viajes, mediana de su tarifa y duracion del viaje
patron_dia = taxis.groupby(['dia_num', 'dia_nombre']).agg(
    viajes=('dia_num', 'size'),mediana_dur=('duracion_min', 'median'),mediana_fare=('fare', 'median'))

distin = taxis.groupby('dia_num')['fecha'].nunique() #cuantas fechas distintas aparece el dia de la semana
patron_dia['promedio_viajes_fecha'] = patron_dia['viajes'] / distin.values #calulamos el promedio de viajes por fecha

print(patron_dia)
```

## Actividad 4. Dataset limpio y comparación (20 puntos)

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
taxis_limpio = taxis[
    (taxis['duracion_min'] >= 1) & #viajes de  1 minuto o menos
    (taxis['duracion_min'] <= 180) &  #quita viajes mas de 180 minutos
    (taxis['velocidad_mph'] <= 60) & #conservamos viajes maximo 60 millas
    (taxis['pickup'].between('2019-03-01', '2019-03-31 23:59:59'))].copy() #nueva base con los filtros

print(f"Total original: {len(taxis)} | Excluidos: {len(taxis) - len(taxis_limpio)} | Retenidos: {len(taxis_limpio)}") #comparamos

print(f"Min duracion: {taxis_limpio['duracion_min'].min()}")
print(f"Max velocidad: {taxis_limpio['velocidad_mph'].max()}")
```

### 4.2 Comparación entre periodos (6 pts)
