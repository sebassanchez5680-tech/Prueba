## Actividad 1. Diagnóstico inicial (20 puntos)

### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
dup_first = hoteles.duplicated(keep='first').sum() #contamos los repetidos unicamente conservando como false la primera aparicion

dup_last = hoteles.duplicated(keep='last').sum() #similar a first pero en este caso la ultima aparicon se le asgina el false

Todo = hoteles.duplicated(keep=False).sum() #aqui contamos a todas por igual, entonces si un valor se repite se marca todas las repeticiones

print(f"Filas keep='first': {dup_first}")
print(f"Filas keep='last': {dup_last}")
print(f"Filas keep=False: {Todo}")
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
proporcion = hoteles.duplicated().mean() #calculamos el porcentaje de sobrantes respecto del total del dataset
conteo_nombre_edad = hoteles.value_counts() #contamos cuantas veces aparecen cada uno de los valores
repetidos = conteo_nombre_edad[conteo_nombre_edad > 1] # contamos a los grupos que aparecen mas de 1 vez

print(f'Porcentaje de filas sobrantes: {proporcion * 100:.2f} %')
print('Grupos repetidos: ', len(repetidos))
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python

hoteles_diag = hoteles.copy() #copia

hoteles_diag['es_duplicado'] = hoteles_diag.duplicated(keep='first') #booleano que identifica las repeticiones posteriores a la primera aparicion

# Agrupamos por hotel para sumar los duplicados y calcular la proporcion (media)
# Con agg aplicamos la operacion para obetener el promedio
duplicados_por_hotel = hoteles_diag.groupby('hotel')['es_duplicado'].agg( Sobrantes='sum',Porcentaje=lambda x: (x.mean() * 100).round(2))

print(duplicados_por_hotel)
print(len(hoteles))
```

### 2.3 ¿En qué se distinguen las filas duplicadas? (6 pts)

Compara las filas involucradas en duplicados (`keep=False`) con el dataset completo en:

- Distribución de `market_segment` (proporciones).
- Distribución de `deposit_type` (proporciones).
- Tasa de cancelación (`is_canceled`).

```python
# Filtramos las filas que están involucradas en algún grupo repetido.
duplicados_involucrados = hoteles[hoteles.duplicated(keep=False)].copy()

# Comparamos las distribuciones como proporciones
market_total = hoteles['market_segment'].value_counts(normalize=True, dropna=False) #contamos los valores de market
market_dup = duplicados_involucrados['market_segment'].value_counts(normalize=True, dropna=False) #contamos los duplicados

deposit_total = hoteles['deposit_type'].value_counts(normalize=True, dropna=False) #hacemos lo mismo para deposit
deposit_dup = duplicados_involucrados['deposit_type'].value_counts(normalize=True, dropna=False)

tasa_cancelacion_total = hoteles['is_canceled'].mean() #calculamos las tasas de desocupacion
tasa_cancelacion_dup = duplicados_involucrados['is_canceled'].mean()

print("Distribución de market_segment:")
print(pd.concat([market_total.rename('total'), market_dup.rename('duplicados')], axis=1))

print("\nDistribución de deposit_type:")
print(pd.concat([deposit_total.rename('total'), deposit_dup.rename('duplicados')], axis=1))

print(f"\nTasa de cancelación - dataset completo: ({tasa_cancelacion_total*100:.2f}%)")
print(f"Tasa de cancelación - filas involucradas en duplicados: ({tasa_cancelacion_dup*100:.2f}%)")
```

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
# Eliminamos duplicados exactos y renumeramos el indice para mantener consistencia (keep='first')
hoteles_limpio = hoteles.drop_duplicates(keep='first', ignore_index=True)

print(f"Filas originales: {len(hoteles)}")
print(f"Filas después de limpieza: {len(hoteles_limpio)}")
print(f"Filas eliminadas: {len(hoteles) - len(hoteles_limpio)}")
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
#metricas para el dataset original.
resumen_original = (hoteles.groupby('hotel').agg(reservas=('hotel', 'size'),tasa_cancelacion=('is_canceled', 'mean'),adr_promedio=('adr', 'mean')).reset_index())

# metricas sin duplicados.
resumen_limpio = (hoteles_limpio.groupby('hotel').agg(reservas=('hotel', 'size'),tasa_cancelacion=('is_canceled', 'mean'),adr_promedio=('adr', 'mean')).reset_index())

print("Resumen original  :")
print(resumen_original)

print("Resumen sin duplicados  :")
print(resumen_limpio)
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# ponemos una etiqueta para distinguir las versiones.
resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'

# unimos los dos conjuntos con pd.concat().
comparacion = (pd.concat([resumen_original, resumen_limpio], ignore_index=True).sort_values(by='hotel').reset_index(drop=True))

print(comparacion)
```
