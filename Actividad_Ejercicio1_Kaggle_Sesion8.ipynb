**Paso 1 y 2 — Configurar credenciales y descargar el dataset.**

```python
# Instala la librería kaggle (si no está instalada ya)
!pip install kaggle --quiet

import os

# Carga el token desde los Secretos de Colab
try:
    from google.colab import userdata
    os.environ["KAGGLE_API_TOKEN"] = userdata.get("KAGGLE_API_TOKEN")
    print("Token de Kaggle cargado")
except ImportError:
    print("Esta celda está diseñada para ejecutarse en Google Colab con el Secreto configurado")

# Completa el identificador del dataset y descárgalo en datos_drug/
dataset_id = "prathamtripathi/drug-classification"  # <- colocamos el nombre del dataset que vamos a buscar

!kaggle datasets download -d {dataset_id} -p datos_drug/ --unzip
```

**Paso 3 — Cargar el archivo descargado en un DataFrame.**

```python
import glob
import pandas as pd

archivos_csv = glob.glob("datos_drug/*.csv")
print(archivos_csv)

# Completa: carga el primer archivo de archivos_csv en un DataFrame llamado df_drug
df_drug = pd.read_csv(archivos_csv[0]) #cargamos el primer csv
```

**Paso 4 — Verificar el resultado.**

```python
print(df_drug.shape) #cuántas filas y columnas llegaron?
print(df_drug.columns) #qué columnas son ?
print(df_drug.dtypes) # de qué tipo es cada una?
print(df_drug.isna().sum()) #hay valores faltantes?

#nos arroja un dataset de 200 lineas y 6 columnas
# las 6 columnas tienen como nombre Age, Sex, BP, Cholesterol, Na_to_K, Drug
#con tipo entero, objeto y flotante
#no hay valores faltantes
```

**Paso 5 — Ficha de procedencia.** Completa cada valor del diccionario (usa el mismo patrón visto en
